```python
import json, pathlib, pycocotools

def read_jsonl(p):
    with open(p) as f:
        for l in f:
            if l.strip():
                yield json.loads(l)

rows = list(read_jsonl("preds.jsonl"))

images, annotations, results = [], [], []
ann_id = 1

for r in rows:
    img_id = int(r["image_id"])
    images.append({"id": img_id})

    for cap in r["references"]:
        annotations.append({
            "image_id": img_id,
            "id": ann_id,
            "caption": cap
        })
        ann_id += 1

    results.append({
        "image_id": img_id,
        "caption": r["prediction"]
    })
```

```python
out = pathlib.Path("metrics")
out.mkdir(exist_ok=True)

json.dump({"images": images, "annotations": annotations},
          open(out / "coco_gt.json", "w"))
json.dump(results, open(out / "coco_preds.json", "w"))

from pycocotools.coco import COCO
from pycocoevalcap.eval import COCOEvalCap

cocoGt = COCO(str(out / "coco_gt.json"))
cocoRes = cocoGt.loadRes(str(out / "coco_preds.json"))

cocoEval = COCOEvalCap(cocoGt, cocoRes)
cocoEval.evaluate()

json.dump(cocoEval.eval,
          open(out / "coco_caption_metrics.json", "w"),
          indent=2)

print(cocoEval.eval)
```
